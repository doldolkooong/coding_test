두 수를 입력받아 두 수의 최대공약수와 최소공배수를 반환하는 함수, solution을 완성해 보세요. 배열의 맨 앞에 최대공약수, 그다음 최소공배수를 넣어 반환하면 됩니다. 예를 들어 두 수 3, 12의 최대공약수는 3, 최소공배수는 12이므로 solution(3, 12)는 [3, 12]를 반환해야 합니다.

제한 사항
두 수는 1이상 1000000이하의 자연수입니다.

입출력 예
n	m	return
3	12	[3, 12]
2	5	[1, 10].

```python
def solution(n, m):
    answer = []

    # 최대공약수를 저장할 변수
    gcd = 0

    # 1부터 n, m 중 더 작은 수까지 확인
    for i in range(1, min(n, m) + 1):

        # i가 n과 m을 둘 다 나누어 떨어지게 하면 공약수
        if n % i == 0 and m % i == 0:
            # 반복할수록 더 큰 공약수로 갱신
            gcd = i

    # 최소공배수
    lcm = n * m // gcd

    # 최대공약수, 최소공배수 순서로 저장
    answer.append(gcd)
    answer.append(lcm)

    return answer
```

```python
n = 3
m = 12

solution(n, m)
```
