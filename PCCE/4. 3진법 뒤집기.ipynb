자연수 n이 매개변수로 주어집니다. n을 3진법 상에서 앞뒤로 뒤집은 후, 이를 다시 10진법으로 표현한 수를 return 하도록 solution 함수를 완성해주세요.

45(10) -> 1200(3) -> 0021 -> 7

```python
def solution(n):
    answer = '' # 뒤집힌 3진수를 저장할 문자열

    while n != 0:
        s = n % 3 # 3으로 나눈 나머지 구하기
        n = n // 3 # 3우로 나는 몫을 n에 재저장
        answer += str(s) # 나머지를 앞에서부터 저장 
    
    return int(answer,3) # 뒤집힌 3진수 문자열을 10진수로 변환
    # 이 문자열을 3진수로 해석하라
```

```python
n = 45
solution(n)
```

```python
n = 125
solution(n)
```
