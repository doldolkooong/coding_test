어떤 문장의 각 알파벳을 일정한 거리만큼 밀어서 다른 알파벳으로 바꾸는 암호화 방식을 시저 암호라고 합니다. 예를 들어 "AB"는 1만큼 밀면 "BC"가 되고, 3만큼 밀면 "DE"가 됩니다. "z"는 1만큼 밀면 "a"가 됩니다. 문자열 s와 거리 n을 입력받아 s를 n만큼 민 암호문을 만드는 함수, solution을 완성해 보세요.

제한 조건
공백은 아무리 밀어도 공백입니다.
s는 알파벳 소문자, 대문자, 공백으로만 이루어져 있습니다.
s의 길이는 8000이하입니다.
n은 1 이상, 25이하인 자연수입니다.

```python
# 1. 결과를 담을 문자열이나 리스트 answer를 만든다.
# 2. 문자열 s를 왼쪽부터 한 글자씩 순회한다.
# 3. 현재 문자가 공백이면 그대로 answer에 넣는다.
# 4. 현재 문자가 대문자인지 소문자인지 구분한다.
# 5. 현재 문자를 알파벳 순서에서 n만큼 뒤로 이동시킨다.
# 6. z나 Z를 넘어가면 다시 a 또는 A부터 이어지도록 처리한다.
# 7. 변환한 문자를 answer에 추가한다.
# 8. 문자열 끝까지 반복한 뒤 answer를 반환한다.

def solution(s, n):
    
    answer = []
    
    # 문자 -> 아스키코드 ord()
    # 아스키코드 -> 문자 chr()
    
    for i in range(len(s)):
        
        # 현재 문자가 공백?
        #   → 그대로 추가하고 continue
        if s[i] == " ":
            answer.append(s[i])
            continue
        # 현재 문자가 소문자?
        #  → 'a' 기준으로 변환
        
        elif 97 <= ord(s[i]) <= 122:
            num = ord(s[i]) - ord('a')   # 0~25로 변환
            num = num + n              # n만큼 이동
            num = num % 26             # 26을 넘으면 다시 처음으로
            num = num + ord('a')       # 아스키 코드값으로 복구
            ch = chr(num)               # 다시 문자로 변환

        # 현재 문자가 대문자?
        #  → 'A' 기준으로 변환
        if 65 <= ord(s[i]) <= 92:
            num = ord(s[i]) - ord('A')   # 0~25로 변환
            num = num + n              # n만큼 이동
            num = num % 26             # 26을 넘으면 다시 처음으로
            num = num + ord('A')       # 아스키 코드값으로 복구
            ch = chr(num)      

        
        answer.append(ch)

    return "".join(answer)
```

```python
solution("AB", 1)
```
